```python
import requests
import os
from dotenv import load_dotenv, find_dotenv

# Force load environment variables from .env file
# override=True ensures we always read the latest values from the file
load_dotenv(find_dotenv(), override=True)

# Get API Key
api_key = os.getenv("OPENWEATHER_API_KEY")

if not api_key:
    print("API key not found. Make sure OPENWEATHER_API_KEY is in your .env file.")
else:
    # Coordinates for New York (Example)
    lat = 40.7128
    lon = -74.0060
    
    url = f"https://api.openweathermap.org/data/2.5/weather?lat={lat}&lon={lon}&appid={api_key}"
    
    # Make the request
    try:
        response = requests.get(url)
        
        print(f"Status Code: {response.status_code}")
        if response.status_code == 200:
            data = response.json()
            print("Weather Data:", data)
        else:
            print("Error:", response.text)
            if response.status_code == 401:
                print("\nNOTE: status 401 usually means the API key is valid but not subscribed to One Call 3.0.")
                print("Try using the 2.5/weather endpoint if you don't have a subscription.")
    except requests.exceptions.RequestException as e:
        print(f"Request failed: {e}")
```
